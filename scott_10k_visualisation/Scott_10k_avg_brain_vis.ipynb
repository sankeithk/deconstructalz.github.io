```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import nilearn, os, pickle
from nilearn import datasets, plotting, maskers
from nilearn.image import new_img_like
from sklearn.preprocessing import MinMaxScaler
from scipy.stats import zscore
import matplotlib.colors
import matplotlib.artist
from matplotlib.colors import ListedColormap
from matplotlib import font_manager

font_path = "/rds/general/user/sk4724/home/arial.ttf"
font_manager.fontManager.addfont(font_path)

plt.rcParams['font.family'] = 'Arial'

def plot_mean_brain(df_path,
                    atlas,
                    plot_stat_map: bool,
                    diagnosis = 3,
                    output_dir='/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/'):
    """
    This function is to plot the average GREY MATTER VALUES of patients from a certain diagnosis

    Arguments accepted.

    df_path: path to a csv file with patient ROI values
    atlas: can be either 'Desikan' or 'Destrieux' (case-insensitive)
    diagnosis: can be either 1 (CN), 2 (MCI), 3 (AD), 'NaN' (unknown diagnosis)
    plot_stat_map: Boolean if we want to plot a nilearn stat map showing subcortical values along an axial slice.
    
    Function to:

    1. Take a dataframe of patient ROI values,
    2. Extract the patient ROI calues with a certain diagnosis,
    3. Get mean ROI values for that diagnosis,
    4. Takes an atlas
    5. Removes any non-GM values from the atlas if needed
    6. Takes the mean values from Step 3 and the cleaned atlas from Step 5 to:
    (a) cortical parcellation (left medial and lateral)
    (b) a nilearn.plotting.plot_stat_map (optional)
    """
    print(f"STARTING FUNCTION...")

    diag_dict = {1: 'Cognitively Normal',
                2: 'Mildly Cognitively Impairment',
                3: "Alzheimer's Disease",
                'NaN': 'Unknown'}
    print(f"\n Printing average ROI values from {atlas.capitalize()} + ASEG parcellation  for Diagnosis = {diag_dict[diagnosis]}")
    print(f"\n Calculating average ROI values...")

    diag_df = pd.read_csv('/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_housekeeping/no_neuroimaging_data_scott10k_alliedhealth.csv', low_memory = False)
    diag_df = diag_df[['DATA_KEY', 'DIAGNOSIS']]

    # If we want to look at ROI values from scans with an unknown diagnosis, we must get mean values with an alternative apprach (you can't input diagnosis = np.nan without an Error)
    if diagnosis == 'NaN':
        print("\n Diagnosis = NaN selected. Trying alternative approach with pd.isna()...")
        diag_df = diag_df[diag_df['DIAGNOSIS'].isna()]
    else: 
        diag_df = diag_df.loc[diag_df['DIAGNOSIS'] == diagnosis]

    #1. Take a dataframe of patient ROI values,
    if (os.path.exists(df_path)) and (df_path.endswith('.csv')):
        roi_df = pd.read_csv(df_path, low_memory = True)
        
        #2. Extract the patient ROI calues with a certain diagnosis,
        merged_df = pd.merge(how = 'left', on = 'DATA_KEY', left = diag_df, right = roi_df)
        print(f"\nNumber of scans involved = {len(merged_df)}")
        merged_df.drop(columns = ['DATA_KEY','DIAGNOSIS'], inplace = True)
        
        #3. Get mean ROI values for that diagnosis,
        print(merged_df.columns.tolist())
        merged_df = merged_df.select_dtypes(include=[np.number])
        merged_df_mean = merged_df.mean()
        
        
        print(merged_df_mean.idxmax(), merged_df_mean.max())

        ##Section to handle the atlas argument: loads a certain atlas depending on the argument passed in the function
        if atlas.lower() == 'desikan':
            atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/MNI152_1mm_desikan_aparc+aseg.nii.gz'
        elif atlas.lower() == 'destrieux':
            atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/MNI152_1mm_destrieux_aparc.a2009+aseg.nii.gz'
        else:
            raise Exception("Please enter either 'desikan' or 'destrieux' (case-insensitive) as arguments for 'atlas'")
            exit()

        ##Clean the atlas of non-GM values (in memory). I'll save this as a GM ROI-only atlas as well to be safe
        if os.path.exists(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz') == False:
            try:
                print(f"\n GM-only atlas for {atlas.capitalize()} parcellation not found - making one...")
                atlas_img = nilearn.image.load_img(atlas_path)
            except Exception as e:
                print(f"Error loading atlas NIfTI file: {e}")
                exit()
        
            lut_dict = pickle.load(open('/rds/general/project/c3nl_scott_students/live/sankeith/standards/lut_dict.pkl','rb'))
            ##lut_dict has the structure of number : roi. We need to reverse that into roi : number.
    
            lut_dict_reversed_keys = lut_dict.values()
            lut_dict_reversed_values = lut_dict.keys()
    
            lut_dict_reversed = dict(zip(lut_dict_reversed_keys, lut_dict_reversed_values))
    
            non_gm_cols = pickle.load(open('/rds/general/project/c3nl_scott_students/live/sankeith/standards/non_gm_cols.pkl','rb'))
    
            non_gm_cols_dict = {}
    
            for col in non_gm_cols:
                if col in lut_dict_reversed.keys():
                    non_gm_cols_dict[col] = lut_dict_reversed[col]

                print(non_gm_cols_dict) #key = non-GM ROI. value = FreeSurfer LUT number for that ROI
                
                atlas_data = atlas_img.get_fdata().astype(int)
        
                non_gm_labels = np.array(list(non_gm_cols_dict.values()))
                
                mask = np.isin(atlas_data, non_gm_labels)
                
                atlas_data[mask] = 0
        
                clean_atlas_img = new_img_like(atlas_img, atlas_data)
        
                clean_atlas_img.to_filename(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz')

        else:
            print(f"\n GM-only atlas for {atlas.capitalize()} parcellation found - using it to plot ROI data...")

            gm_atlas_img = nilearn.image.load_img(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz')

            merged_df_mean_array = merged_df_mean.to_numpy()

            ##Use nilearn's NiftiLabelsMasker to take our 1D array of mean ROI values and map them to the atlas
            labels_masker = maskers.NiftiLabelsMasker(labels_img=gm_atlas_img, standardize=None)

            ##Labels masker now learns what ROIs there are in a certain atlas
            labels_masker.fit()
            
            print(f"\n Mean values for {diag_dict[diagnosis]} are being mapped onto the {atlas.capitalize()} + ASEG atlas...")

            ## Average ROI values are now matched to the atlas ROIs. NB: MAKE SURE YOUR ROIs' FREESURFER IDs GO UP IN ASCENDING ORDER BEFORE THIS STEP!
            stat_img = labels_masker.inverse_transform(merged_df_mean_array.reshape(1,-1))

            img_data = stat_img.get_fdata()

            non_zero_mask = img_data != 0
            
            img_data[non_zero_mask] = img_data[non_zero_mask].reshape(-1, 1).flatten()

            scaled_stat_img = nilearn.image.new_img_like(stat_img, img_data)

            fsaverage = datasets.fetch_surf_fsaverage()
            cmap = plt.get_cmap('coolwarm', 8)

            if diagnosis == np.nan:
                diagnosis = 'nan'

            print(f"\n Plotting cortical values...")
            
            # Use the plot_img_on_surf function
            plotting.plot_img_on_surf(
                stat_map = scaled_stat_img,
                mask_img=None,
                views=['lateral', 'medial'],
                hemispheres=['left'],
                title=f'GM ROI: {atlas.capitalize()}, diagnosis = {diagnosis}',
                colorbar=True,
                vmin=-1,
                vmax=1,
                cmap='coolwarm'
            )
            
            print(f"\n Saving figure...")
            plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_gm_lh_surface_view.png", dpi=600)
            plt.show()

            if plot_stat_map:
                    print(f"\n Plotting stat map...")

                    plotting.plot_stat_map(
                    stat_img,
                    display_mode='z',
                    vmin=-1,
                    vmax=1,
                    cut_coords=[0],
                    cmap=cmap,
                    bg_img=None
                )
                    print(f"\n Saving figure...")
                    plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_gm_stat_map.png", dpi=600)
                    plt.show()

roi_csv_list = ['/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_housekeeping/gm_only_scraped_desikan_stats.csv',
                '/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_housekeeping/gm_only_scraped_destrieux_stats.csv']

diagnoses = [1,2,3, 'NaN']

for file in roi_csv_list:
    basename = file.split('/')[-1]
    atlas = basename.split('_')[-2]
    for diag in diagnoses:
        plot_mean_brain(file, atlas = atlas, diagnosis = diag, plot_stat_map = True)
    
    
```

```python
font_path = "/rds/general/user/sk4724/home/arial.ttf"
font_manager.fontManager.addfont(font_path)

def plot_mean_brain_nan(df,
                    atlas,
                    plot_stat_map: bool,
                    diagnosis = 'NaN',
                    output_dir='/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/'):
    """
    A function to strictly visualise the CLEANED average VBM ROI values for NaN patients.

    Arguments accepted.

    df: path to a csv file with patient ROI values
    atlas: can be either 'Desikan' or 'Destrieux' (case-insensitive)
    diagnosis: NaN only (duh)
    plot_stat_map: Boolean if we want to plot a nilearn stat map showing subcortical values along an axial slice.
    
    Function to:

    1. Take a dataframe of patient ROI values,
    2. Extract the patient ROI calues with a certain diagnosis,
    3. Get mean ROI values for that diagnosis,
    4. Takes an atlas
    5. Removes any non-GM values from the atlas if needed
    6. Takes the mean values from Step 3 and the cleaned atlas from Step 5 to:
    (a) cortical parcellation (left medial and lateral)
    (b) a nilearn.plotting.plot_stat_map (optional)
    """
    print(f"STARTING FUNCTION...")

    diag_dict = {'NaN': 'Unknown'}
    print(f"\n Printing average ROI values from {atlas.capitalize()} + ASEG parcellation  for Diagnosis = {diag_dict[diagnosis]}")
    print(f"\n Calculating average ROI values...")

    mean_df = df.mean()
    print(mean_df)
    
    print(mean_df.idxmax(), mean_df.max())

    ##Section to handle the atlas argument: loads a certain atlas depending on the argument passed in the function
    if atlas.lower() == 'desikan':
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/MNI152_1mm_desikan_aparc+aseg.nii.gz'
    elif atlas.lower() == 'destrieux':
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/MNI152_1mm_destrieux_aparc.a2009+aseg.nii.gz'
    else:
        raise Exception("Please enter either 'desikan' or 'destrieux' (case-insensitive) as arguments for 'atlas'")
        exit()

    ##Clean the atlas of non-GM values (in memory). I'll save this as a GM ROI-only atlas as well to be safe
    if os.path.exists(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz') == False:
        try:
            print(f"\n GM-only atlas for {atlas.capitalize()} parcellation not found - making one...")
            atlas_img = nilearn.image.load_img(atlas_path)
        except Exception as e:
            print(f"Error loading atlas NIfTI file: {e}")
            exit()
        
        lut_dict = pickle.load(open('/rds/general/project/c3nl_scott_students/live/sankeith/standards/lut_dict.pkl','rb'))
        ##lut_dict has the structure of number : roi. We need to reverse that into roi : number.

        lut_dict_reversed_keys = lut_dict.values()
        lut_dict_reversed_values = lut_dict.keys()

        lut_dict_reversed = dict(zip(lut_dict_reversed_keys, lut_dict_reversed_values))

        non_gm_cols = pickle.load(open('/rds/general/project/c3nl_scott_students/live/sankeith/standards/non_gm_cols.pkl','rb'))

        non_gm_cols_dict = {}
    
        for col in non_gm_cols:
            if col in lut_dict_reversed.keys():
                non_gm_cols_dict[col] = lut_dict_reversed[col]

            print(non_gm_cols_dict) #key = non-GM ROI. value = FreeSurfer LUT number for that ROI
            
            atlas_data = atlas_img.get_fdata().astype(int)
    
            non_gm_labels = np.array(list(non_gm_cols_dict.values()))
            
            mask = np.isin(atlas_data, non_gm_labels)
            
            atlas_data[mask] = 0
    
            clean_atlas_img = new_img_like(atlas_img, atlas_data)
    
            clean_atlas_img.to_filename(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz')

    else:
        print(f"\n GM-only atlas for {atlas.capitalize()} parcellation found - using it to plot ROI data...")

        gm_atlas_img = nilearn.image.load_img(f'/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_{atlas.lower()}+aseg.nii.gz')

        mean_df_arr = mean_df.to_numpy()

        ##Use nilearn's NiftiLabelsMasker to take our 1D array of mean ROI values and map them to the atlas
        labels_masker = maskers.NiftiLabelsMasker(labels_img=gm_atlas_img, standardize=None)

        ##Labels masker now learns what ROIs there are in a certain atlas
        labels_masker.fit()
        
        print(f"\n Mean values for {diag_dict[diagnosis]} are being mapped onto the {atlas.capitalize()} + ASEG atlas...")

        ## Average ROI values are now matched to the atlas ROIs. NB: MAKE SURE YOUR ROIs' FREESURFER IDs GO UP IN ASCENDING ORDER BEFORE THIS STEP!
        stat_img = labels_masker.inverse_transform(mean_df_arr.reshape(1,-1))

        img_data = stat_img.get_fdata()

        non_zero_mask = img_data != 0
        
        img_data[non_zero_mask] = img_data[non_zero_mask].reshape(-1, 1).flatten()

        scaled_stat_img = nilearn.image.new_img_like(stat_img, img_data)

        fsaverage = datasets.fetch_surf_fsaverage()
        cmap = plt.get_cmap('coolwarm', 8)

        if diagnosis == np.nan:
            diagnosis = 'nan'

        print(f"\n Plotting cortical values...")
        
        # Use the plot_img_on_surf function
        plotting.plot_img_on_surf(
            stat_map = scaled_stat_img,
            mask_img=None,
            views=['lateral', 'medial'],
            hemispheres=['left'],
            title=f'GM ROI: {atlas.capitalize()}, diagnosis = {diagnosis} - cleaned',
            colorbar=True,
            vmin=-1,
            vmax=1,
            cmap='coolwarm'
        )
            
        print(f"\n Saving figure...")
        plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_cleaned_gm_lh_surface_view.png", dpi=600)
        plt.show()

        if plot_stat_map:
                print(f"\n Plotting stat map...")

                plotting.plot_stat_map(
                stat_img,
                display_mode='z',
                vmin=-1,
                vmax=1,
                cut_coords=[0],
                cmap=cmap,
                bg_img=None
            )
                print(f"\n Saving figure...")
                plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_cleaned_gm_stat_map.png", dpi=600)
                plt.show()

for df, atlas in [(destrieux_df_onlynan, 'destrieux'), (desikan_df_onlynan, 'desikan')]:
        plot_mean_brain_nan(df, atlas = atlas, plot_stat_map = True)
    
    
```

```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import nilearn, os, pickle
from nilearn import datasets, plotting, maskers
from nilearn.image import new_img_like
from sklearn.preprocessing import MinMaxScaler
from scipy.stats import zscore
import matplotlib.colors
import matplotlib.artist
from matplotlib.colors import ListedColormap
from matplotlib import font_manager

font_path = "/rds/general/user/sk4724/home/arial.ttf"
font_manager.fontManager.addfont(font_path)

plt.rcParams['font.family'] = 'Arial'

def plot_mean_brain_sbm(df,
                    atlas,
                    plot_stat_map: bool,
                    cortical: bool,
                    diagnosis = 3,
                    output_dir='/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/'):
    """
    This function is to plot the average SBM ROI GMVs of patients from a certain diagnosis

    Arguments accepted.

    df: pre-processed dataframe. Needs ONLY diagnosis and patient ROI values
    atlas: can be either 'Desikan' or 'Destrieux' (case-insensitive)
    diagnosis: can be either 1 (CN), 2 (MCI), 3 (AD), 'NaN' (unknown diagnosis)
    plot_stat_map: Boolean if we want to plot a nilearn stat map showing subcortical values along an axial slice.
    cortical: Boolean if the data is cortical thickness or not.
    
    Function to:

    1. Take a dataframe of patient ROI values,
    2. Extract the patient ROI calues with a certain diagnosis,
    3. Get mean ROI values for that diagnosis,
    4. Takes an atlas
    5. Removes any non-GM values from the atlas if needed
    6. Takes the mean values from Step 3 and the cleaned atlas from Step 5 to:
    (a) cortical parcellation (left medial and lateral)
    (b) a nilearn.plotting.plot_stat_map (optional)
    """
    print(f"STARTING FUNCTION...")

    # Debug step - prints out the arguments being used
    print(f'Variables being used: {df.columns.tolist()}')
    print(f"\nAtlas = {atlas}")
    print(f"\nDiagnosis = {diagnosis}")

    diag_dict = {1: 'Cognitively Normal',
                2: 'Mildly Cognitively Impairment',
                3: "Alzheimer's Disease",
                'NaN': 'Unknown'}
    print(f"\n Printing average ROI values from {atlas.capitalize()} + ASEG parcellation  for Diagnosis = {diag_dict[diagnosis]}")
    print(f"\n Calculating average ROI values...")

    # If we want to look at ROI values from scans with an unknown diagnosis, we must get mean values with an alternative apprach (you can't input diagnosis = np.nan without an Error)
    if diagnosis == 'NaN':
        print("\n Diagnosis = NaN selected. Trying alternative approach with pd.isna()...")
        
        bool_mask = pd.isnull(df['DIAGNOSIS'])
        df = df[bool_mask]
        
    else: 
        df = df[df['DIAGNOSIS'] == diagnosis]

    #1. Take a dataframe of patient ROI values,
    cleaned_df = df.drop(columns = 'DIAGNOSIS')
        
        #3. Get mean ROI values for that diagnosis,
    merged_df_mean = cleaned_df.mean()
    print(f"Mean values: \n{merged_df_mean}")

        ##Section to handle the atlas argument: loads a certain atlas depending on the argument passed in the function
    if atlas.lower() == 'desikan' and cortical == False:
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_desikan+aseg.nii.gz'
    elif atlas.lower() == 'desikan' and cortical:
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/cortical_only_MNI152_1mm_desikan+aseg.nii.gz'
    elif atlas.lower() == 'destrieux' and cortical == False:
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/gm_only_MNI152_1mm_destrieux+aseg.nii.gz'
    elif atlas.lower() == 'destrieux' and cortical:
        atlas_path = '/rds/general/project/c3nl_scott_students/live/sankeith/standards/cortical_only_MNI152_1mm_destrieux+aseg.nii.gz'
    else:
        raise Exception("Please enter either 'desikan' or 'destrieux' (case-insensitive) as arguments for 'atlas'")
        exit()

    if cortical:
        print(f"\n Cortex-only atlas for {atlas.capitalize()} parcellation requested - using it to plot ROI data...")
    else:
        print(f"\n Cortical and subcortical atlas for {atlas.capitalize()} parcellation requested - using it to plot ROI data...")
    
    print(f"Atlas = {atlas_path}")

    gm_atlas_img = nilearn.image.load_img(atlas_path)

    merged_df_mean_array = merged_df_mean.to_numpy()

    ##Use nilearn's NiftiLabelsMasker to take our 1D array of mean ROI values and map them to the atlas
    labels_masker = maskers.NiftiLabelsMasker(labels_img=gm_atlas_img, standardize=None)

    ##Labels masker now learns what ROIs there are in a certain atlas
    labels_masker.fit()
            
    print(f"\n Mean values for {diag_dict[diagnosis]} are being mapped onto the {atlas.capitalize()} + ASEG atlas...")

    ## Average ROI values are now matched to the atlas ROIs. NB: MAKE SURE YOUR ROIs' FREESURFER IDs GO UP IN ASCENDING ORDER BEFORE THIS STEP!
    stat_img = labels_masker.inverse_transform(merged_df_mean_array.reshape(1,-1))

    img_data = stat_img.get_fdata()

    non_zero_mask = img_data != 0
            
    img_data[non_zero_mask] = img_data[non_zero_mask].reshape(-1, 1).flatten()

    scaled_stat_img = nilearn.image.new_img_like(stat_img, img_data)

    fsaverage = datasets.fetch_surf_fsaverage()
    cmap = plt.get_cmap('coolwarm', 8)

    if diagnosis == np.nan:
        diagnosis = 'nan'

    print(f"\n Plotting cortical values...")

    if cortical:
        title = f'Cortical Thickness: {atlas.capitalize()}, diagnosis = {diag_dict[diagnosis]}'
    else:
        title = f'SBM ROI GMV: {atlas.capitalize()}, diagnosis = {diag_dict[diagnosis]}'
    # Use the plot_img_on_surf function
    plotting.plot_img_on_surf(
        stat_map = scaled_stat_img,
        mask_img=None,
        views=['lateral', 'medial'],
        hemispheres=['left'],
        title= title,
        colorbar=True,
        vmin=-1,
        vmax=1,
        cmap='coolwarm'
    )
            
    print(f"\n Saving figure...")
    if cortical:
        plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_ct_lh_surface_view.png", dpi=600)
    else:
        plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_gmv_roi_lh_surface_view.png", dpi=600)

    plt.show()

    if plot_stat_map:
        print(f"\n Plotting stat map...")

        plotting.plot_stat_map(
        stat_img,
        display_mode='z',
        vmin=-1,
        vmax=1,
        cut_coords=[0],
        cmap=cmap,
        bg_img=None)
        
        print(f"\n Saving figure...")

        if cortical:
            plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_ct_stat_map.png", dpi=600)
            plt.show()
        else:
            plt.savefig(f"/rds/general/project/c3nl_scott_students/live/sankeith/scott_10k_visualisation/avg_brain_vis/{atlas.lower()}_{diagnosis}_gmv_roi_stat_map.png", dpi=600)
            plt.show()

roi_csv_list = [(desikan_cort_df, 'desikan', True), (desikan_roi_gmv_df, 'desikan', False), (destrieux_cort_df, 'destrieux', True) , (destrieux_roi_gmv_df, 'destrieux', False)]

diagnoses = [1,2,3, 'NaN']

for file, atlas, cort in roi_csv_list:
    for diag in diagnoses:
        plot_mean_brain_sbm(file, atlas = atlas, diagnosis = diag, plot_stat_map = True, cortical = cort)
```
